```python
## Function to get channel statistics
def get_channel_stats(youtube, channel_ids):
    all_data = []
    request = youtube.channels().list(
                part='snippet,contentDetails,statistics',
                id=','.join(channel_ids))
    response = request.execute() 
    
    for i in range(len(response['items'])):
        data = dict(Channel_Name = response['items'][i]['snippet']['title'],
                    Subscribers = response['items'][i]['statistics']['subscriberCount'],
                    Views = response['items'][i]['statistics']['viewCount'],
                    Total_Videos = response['items'][i]['statistics']['videoCount'],
                    playlist_id = response['items'][i]['contentDetails']['relatedPlaylists']['uploads'],
                    Created_Date=response['items'][i]['snippet']['publishedAt'])
        all_data.append(data)
    
    return all_data

##Extract data using the function above
channel_statistics = get_channel_stats(youtube, channel_ids)

##Create panda dataframe, preprocess, and sort by subscriber
channel_data = pd.DataFrame(channel_statistics)
   
    #change type of column for data enginieering later
channel_data ['Subscribers']= channel_data ['Subscribers'].astype('int')
channel_data ['Created_Date'] = pd.to_datetime(channel_data['Created_Date'])
channel_data ['Views']= channel_data ['Views'].astype('int')
channel_data ['Total_Videos']= channel_data ['Total_Videos'].astype('int')
    #data enginieer
channel_data['Subscribers per video']=round(channel_data['Subscribers'] / channel_data['Total_Videos'])
channel_data['Views per video']=round(channel_data['Views'] / channel_data['Total_Videos'])
    
    #sort dataframe and display data with comma seperator
channel_data=channel_data.sort_values(by='Subscribers',ascending=False).reset_index(drop=True)
format_dict = {'Subscribers':'{:,.0f}', 
                'Views':'{:,.0f}', 
                'Total_Videos':'{:,.0f}',
                'Subscribers per video':'{:,.0f}',
                'Views per video':'{:,.0f}'}
channel_data.style.format(format_dict)
```

```python
# Construct a datafram containing all records for all 20 channels
channel_names= channel_data['Channel_Name']

def combine_df(channel_names):
    dfs=[]
    for name in channel_names:
        df = pd.read_csv(f'channel_data/{name}.csv')
        dfs.append(df)
    return dfs

channels_df = pd.concat(combine_df(channel_names), ignore_index=True).drop(columns='Unnamed: 0')
```

```python
####Preprocessing: calculate video length, convert datetype, and split tag keywords and title keywords
def Processing_df (df):
    
    #calulate video length in seconds
    def time_to_seconds(time_str):
        # Check if the input contains minutes and seconds (xMxS format)
        match = re.match(r'(\d+)M(\d+)S', time_str)
        if match:
            minutes = int(match.group(1))
            seconds = int(match.group(2))
            total_seconds = minutes * 60 + seconds
            return total_seconds
        else:
            # Check if the input contains only seconds (xS format)
            match = re.match(r'(\d+)S', time_str)
            if match:
                seconds = int(match.group(1))
                return seconds
            else:
                match=re.match(r'(\d+)M',time_str)
                if match:
                    minutes=int(match.group(1))
                    return minutes*60
                else:
                    return 0
    
    # Use defined function to calculate video length in seconds and do some other processing
    df['Duration_Seconds'] = df['Duration'].apply(time_to_seconds)
    df['Tag_Keywords'] = df['Tags_combined'].apply(lambda x: [tag.strip() for tag in x.split(',')])
    df['Published_Time'] = pd.to_datetime(df['Published_Time'])
    
    #extract keywords from title
    nltk.download('stopwords')
    nltk.download('punkt')
    def preprocess_text(text):
        text = re.sub(r'#\w+', '', text) # delete '#'
        text = re.sub(r'[^\W\S]', '', text) # delete all punction
        words = nltk.word_tokenize(text) # split words
        words = [word for word in words if word.lower() not in set(stopwords.words('english'))] #delete stopwords
        return " ".join(words)
    df['Cleaned_Title'] = df['Title'].apply(preprocess_text)

    return df

channels_df = Processing_df (channels_df)
```

```python
ellie=pd.read_csv('/kaggle/input/popular-youtube-channels-on-golden-retrievers/Archive/Ellie Golden Life.csv').drop(columns='Unnamed: 0')

#calulate video length in seconds
def time_to_seconds(time_str):
    # Check if the input contains minutes and seconds (xMxS format)
    match = re.match(r'(\d+)M(\d+)S', time_str)
    if match:
        minutes = int(match.group(1))
        seconds = int(match.group(2))
        total_seconds = minutes * 60 + seconds
        return total_seconds
    else:
        # Check if the input contains only seconds (xS format)
        match = re.match(r'(\d+)S', time_str)
        if match:
            seconds = int(match.group(1))
            return seconds
        else:
            match=re.match(r'(\d+)M',time_str)
            if match:
                minutes=int(match.group(1))
                return minutes*60
            else:
                # raise ValueError("Invalid time format. Expected 'xM'/'xMxS'/'xS' format.")
                return 0

ellie['Duration_Seconds'] = ellie['Duration'].apply(time_to_seconds)

ellie['Tag_Keywords'] = ellie['Tags_combined'].apply(lambda x: [tag.strip() for tag in x.split(',')])

ellie['Published_Time'] = pd.to_datetime(ellie['Published_Time'])

display(ellie.head(3))
```

```python
#Calculate number of rows for 20% of records
performance_index=int(np.ceil(ellie.shape[0]*0.2))

#sort df by views, likes, and comments
ellie=ellie.sort_values(by=['Views','Likes','Comments'],ascending=False)

#Extract top 20% and bottom 20%
top_ellie=ellie.head(performance_index)
# display(top_ellie.head(3))
bottom_ellie=ellie.tail(performance_index)
# display(bottom_ellie.head(3))

# Calculate means and standard deviations for each set
top_stats = top_ellie.mean(numeric_only=True), top_ellie.std(numeric_only=True)
bottom_stats = bottom_ellie.mean(numeric_only=True), bottom_ellie.std(numeric_only=True)

# Plot means and standard deviations
columns = ['Views', 'Likes', 'Comments', 'Duration_Seconds']

# Create subplots
fig, axes = plt.subplots(2, 2, figsize=(12, 8))
fig.suptitle('Means and Std Dev for Different Columns')

for i, ax in enumerate(axes.ravel()):
    col = columns[i]
    means = [top_stats[0][col], bottom_stats[0][col]]
    stds = [top_stats[1][col], bottom_stats[1][col]]
    
    bars = ax.bar(['Top Group', 'Bottom Group'], means, yerr=stds, capsize=10)
    
    for bar, mean, std in zip(bars, means, stds):
        ax.text(bar.get_x() + bar.get_width() / 2, bar.get_height() + 10, f'Mean: {mean:.2f}\nStd: {std:.2f}', ha='left', va='bottom')
    
    ax.set_title(f'{col}: Mean and Std Dev')
    ax.set_ylabel(col)

# Adjust subplot layout
plt.tight_layout()
plt.subplots_adjust(top=0.85)
plt.show()
```

```python
api_key = 'Use your own API key'
youtube = googleapiclient.discovery.build('youtube', 'v3', developerKey=api_key)

# Define the maximum number of results to fetch per request
max_results_per_request = 50  # The maximum allowed by the youtube API

# Define your search keywords as a regex pattern
search_pattern = r'puppies|puppy|dog|dogs|golden|golden retriever|retriever|canine|pet|furry friend' 

# Initialize variables to keep track of the total videos and the videos matching the criteria
total_videos = 0
matching_videos = []

# Continue fetching videos until reaching 1000 matching videos or there are no more results
while total_videos < 1000:
    # Calculate the remaining number of videos to fetch in this request
    remaining_videos_to_fetch = 1000 - total_videos
    num_videos_to_fetch = min(remaining_videos_to_fetch, max_results_per_request)

    # Call the API to search for videos with a maximum of 'num_videos_to_fetch'
    search_response = youtube.search().list(
        q=search_pattern,
        type='video',
        part='snippet',
        maxResults=num_videos_to_fetch,
        pageToken=None if total_videos == 0 else next_page_token  # Use 'pageToken' for pagination
    ).execute()

    # Fetch details for each video and add them to the matching_videos list
    for item in search_response['items']:
        video_info = item['snippet']
        video_id = item['id']['videoId']
        
        # Extract video details
        video_title = video_info.get('title', '')
        video_description = video_info.get('description', '')
        video_tags = video_info.get('tags', [])
        published_time = video_info.get('publishedAt', '')
        
        # Extract channel details
        channel_name = video_info['channelTitle']
        channel_id = video_info['channelId']

        # Call the API to fetch video statistics
        video_stats_response = youtube.videos().list(
            part='statistics,contentDetails',
            id=video_id
        ).execute()

        # Extract statistics and duration information
        video_stats = video_stats_response['items'][0]['statistics']
        video_content_details = video_stats_response['items'][0]['contentDetails']
        
        views = int(video_stats.get('viewCount', 0))
        likes = int(video_stats.get('likeCount', 0))
        comments = int(video_stats.get('commentCount', 0))
        video_duration = video_content_details['duration']

        matching_videos.append({
            'Video_ID': video_id,
            'Title': video_title,
            'Description': video_description if video_description else ' ',
            'Tags': ' '.join(video_tags) if video_tags else ' ',
            'PublishedTime': published_time,
            'Views': views,
            'Likes': likes,
            'Comments': comments,
            'Duration': video_duration,
            'ChannelName': channel_name,
            'ChannelID': channel_id
        })

    # Update the total number of videos fetched
    total_videos += len(search_response['items'])

    # Check if there are more pages of results
    next_page_token = search_response.get('nextPageToken')

    # If there are no more pages or reached desired number of videos, break the loop
    if not next_page_token or total_videos >= 1000:
        break

# Create a DataFrame from the list of matching videos
matching_video_data_df = pd.DataFrame(matching_videos)

# Display the DataFrame as a table
print(matching_video_data_df.shape)
```

```python
####Preprocessing: calculate video length, convert datetype, and split tag keywords and title keywords
def Processing_df (df):
    
    #calulate video length in seconds
    def time_to_seconds(time_str):
        # Check if the input contains minutes and seconds (xMxS format)
        match = re.match(r'PT(\d+)M(\d+)S', time_str)
        if match:
            minutes = int(match.group(1))
            seconds = int(match.group(2))
            total_seconds = minutes * 60 + seconds
            return total_seconds
        else:
            # Check if the input contains only seconds (xS format)
            match = re.match(r'PT(\d+)S', time_str)
            if match:
                seconds = int(match.group(1))
                return seconds
            else:
                match=re.match(r'PT(\d+)M',time_str)
                if match:
                    minutes=int(match.group(1))
                    return minutes*60
                else:
                    return 0
    
    # Use defined function to calculate video length in seconds and do some other processing
    df['Duration_Seconds'] = df['Duration'].apply(time_to_seconds)
    df['Tag_Keywords'] = df['Tags'].apply(lambda x: [tag.strip() for tag in x.split(',')])
    df['PublishedTime'] = pd.to_datetime(df['PublishedTime'])
    
    #extract keywords from title
    nltk.download('stopwords')
    nltk.download('punkt')
    def preprocess_text(text):
        text = re.sub(r'#\w+', '', text) # delete '#'
        text = re.sub(r'[^\W\S]', '', text) # delete all punction
        words = nltk.word_tokenize(text) # split words
        words = [word for word in words if word.lower() not in set(stopwords.words('english'))] #delete stopwords
        return " ".join(words)
    df['Cleaned_Title'] = df['Title'].apply(preprocess_text)

    return df

matching_video_data_df=Processing_df(matching_video_data_df)
```

```python
#Create Shorts DF by applying filters using 3 criterias: duration <=60seconds, #views >=1000, and published within 2wks

def create_shorts_df(df):
    
    #Filter videos with duration less than 60 seconds and views number greater than 1000
    shorts_df=df.copy()
    shorts_df=shorts_df[shorts_df['Duration_Seconds']<=60]
    shorts_df=shorts_df[shorts_df['Views']>=1000]

    #Filter shorts for creation dates within 2 weeks from now
    current_date = datetime.now()
    two_weeks_ago = current_date - timedelta(weeks=2)
    published_after_date = two_weeks_ago.strftime('%Y-%m-%dT%H:%M:%SZ')

    shorts_df=shorts_df[shorts_df['PublishedTime']>=published_after_date].sort_values(by='Views',ascending=False).reset_index(drop=True)
    
    return shorts_df

matching_shorts_data_df=create_shorts_df(matching_video_data_df)
print(matching_shorts_data_df.shape)
```
